```python
def mul_upto_three(*args) -> float:
    """Return the multiplication of up to three numbers"""
    if len(args) < 1 or len(args) > 3:
        return "Неправильное кол-во элементов"
    ans = 1
    for el in args:
        ans *= el
    return ans
a1 = 2
a2 = 3
b1 = 16.2
b2 = 188.4
b3 = 2
c1 = 12
print(mul_upto_three(a1,a2))
print(mul_upto_three(b1,b2))
print(mul_upto_three(c1))
```
