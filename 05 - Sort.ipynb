Написать 4 функции порождения списка целых значений: отсортированного (по возрастанию и убыванию), случайного и почти отсортированного по возрастанию. Каждая функция принимает длину списка в качестве аргумента.

```python
from random import randint

def sort_arr(n):
    lst = random_arr(n)
    lst.sort()
    return lst


def sortrev_arr(n):
    lst = random_arr(n)
    lst.sort(reverse=True)
    return lst


def random_arr(n):
    return [randint(0,100) for _ in range(n)]

def almstsort_arr(n):
    lst = sort_arr(n)
    for i in range(0, n-1,2):
        lst[i], lst[i+1] = lst[i+1], lst[i]
    return lst

print(f'возрастаниe {sort_arr(20)}')
print(f'убываниe {sortrev_arr(10)}')
print(f'случайный {random_arr(10)}')
print(f'почти сортированный   {almstsort_arr(20)}')
```

Для каждой длины списка функция тестирования возвращает минимальное, среднее и наибольшее количество сравнений и перестановок.

```python
def bubble_sort(lst2):
    lst = lst2.copy()
    swap = 0
    compare = 0
    n = len(lst)
    for i in range(n-1, 0, -1):
        for j in range(i):
            compare += 1
            if lst[j] > lst[j+1]:
                swap += 1
                lst[j], lst[j+1] = lst[j+1], lst[j]

    return lst, compare, swap
```

```python

def stone_bubble_sort(lst2):
    comps = 0
    swaps = 0
    lst = lst2.copy()
    for j in range(len(lst)-1):  # опускаем камешек
        comps += 1
        if lst[j] > lst[j+1]:  # больший элемент двигаем ближе к концу списка
            swaps += 1
            lst[j+1], lst[j] = lst[j], lst[j+1]
            for i in range(j, 0, -1):  # поднимаем пузырек
                comps += 1
                if lst[i-1] > lst[i]:  # меньший элемент двигаем ближе к началу списка
                    swaps += 1
                    lst[i], lst[i-1] = lst[i-1], lst[i]
    return lst, comps, swaps
    
lst = [1, 12, 10, 0, 122, 123, 34, 55, 34, 11, 6]
lst_new, comps, swaps = stone_bubble_sort(lst)
print(lst)
```

Реализовать быструю сортировку с возможностью подсчета операций. Сравнить производительность алгоритма на случайных и упорядоченных (и почти упорядоченных) данных.

 quick sort предполагает деление массива на две части, в одной из которых находятся элементы меньше определённого значения, в другой – больше или равные.

```python
def quick_sort(lst: list, count: list):
    if len(lst) <= 1:
        count[0] += 1
        return lst, count

    separator = lst[len(lst) // 2]
    lesser_lst = []
    equal_lst = []
    greater_lst = []

    for el in lst:
        count[0] += 1
        if el < separator:
            count[0] += 1
            lesser_lst.append(el)
        elif el == separator:
            count[0] += 2
            equal_lst.append(el)
        else:
            count[0] += 3
            greater_lst.append(el)

    tmp_lst1, _ = quick_sort(lesser_lst, count)
    tmp_lst2, _ = quick_sort(greater_lst, count)
    return tmp_lst1 + equal_lst + tmp_lst2, count
```

```python
lst = [1, 10, 11, 2, 12, 122, 44, 34]
lst_sorted, count = quick_sort(lst, [0])
print(lst_sorted, *count)

print('элементов,  random_arr, almost_sorted_arr, sorted_arr, sortedrev_arr')
for i in range(100, 10000, 1000):
    lst_sorted1, count1 = quick_sort(random_arr(i), [0])
    lst_sorted2, count2 = quick_sort(almstsort_arr(i), [0])
    lst_sorted3, count3 = quick_sort(sortrev_arr(i), [0])
    lst_sorted4, count4 = quick_sort(sort_arr(i), [0])

    print(i, count1, count2, count3, count4)
```

Улучшить работу быстрой сортировки на упорядоченных и почти упорядоченных за счет изменения алгоритма выбора элемента для разделения массива.

```python
import statistics
```

```python
def modified_qsort(lst):
    if len(lst) <= 1:
        return lst
    else:
        mediana = statistics.median([lst[0], lst[len(lst)//2], lst[-1]]) 
        less = [x for x in lst if x < mediana ]
        equal = [x for x in lst if x == mediana ]
        greater = [x for x in lst if x > mediana ]
        return modified_qsort(less) + equal + modified_qsort(greater)
```

Реализовать эффективный алгоритм получения из большого количества отсортированных списков одного отсортированного списка.

```python
def merge(nlist_1, nlist_2):
    list_1 = nlist_1.copy()
    list_2 = nlist_2.copy()
    sorted_list = []
    while list_1 and list_2 :
        min_el_1 = min(list_1)
        min_el_2 = min(list_2)
        if min_el_1 < min_el_2:
            sorted_list.append(min_el_1)
            list_1.remove(min_el_1)
        else:
            sorted_list.append(min_el_2)
            list_2.remove(min_el_2)
    if list_1:
        sorted_list = sorted_list + list_1
    if list_2:
        sorted_list = sorted_list + list_2

    return sorted_list

def merge_lists(lists):
    res = []
    for lst in lists:
        res = merge(res, lst)
    return res
```

На 2[𝑛/2𝑘]+1

т.е. при первой сортировке будут использоваться шаги: 2[n/4]+1, 2[n/8]+1, ... , 3, 1 .

```python
from random import randint
import time
from math import floor



def gap_insertion_sort(a_list, start, gap):
    for i in range(start + gap, len(a_list), gap):
        current_value = a_list[i]
        position = i
        while position >= gap and a_list[position - gap] > current_value:
            a_list[position] = a_list[position - gap]
            position = position - gap
        a_list[position] = current_value


def shell_sort_mod(a_list):
    temp = 1
    sublist_count = 2 * floor(len(a_list)/(2**(temp+1)))+1
    count = 0
    while True:
        for start_position in range(sublist_count):
            gap_insertion_sort(a_list, start_position, sublist_count)
        print("After inc. of size", sublist_count, "Lst:", a_list)
        temp += 1
        count += 1
        if sublist_count == 1:
            break
        sublist_count = 2 * floor(len(a_list)/(2**(temp+1)))+1
    return count


def shell_sort(a_list):
    sublist_count = len(a_list)//2
    count = 0
    while sublist_count > 0:
        for start_position in range(sublist_count):
            gap_insertion_sort(a_list, start_position, sublist_count)
        print("After inc. of size", sublist_count, "Lst:", a_list)
        count += 1
        sublist_count //= 2
    return count


l = [randint(1, 1000) for i in range(100)]
b = l[::]
print(l, '\n')
print(shell_sort_mod(l),'\n')
print(b,'\n')
print(shell_sort(b),'\n')
```
