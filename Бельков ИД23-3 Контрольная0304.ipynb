1. Создайте класс Length (Длина), имеющий свойства

• value (значение),

• unit (единица измерения).

При изменении единицы измерения значение должно соответственно меняться. Например, при переходе от сантиметров к метрам значение должно уменьшаться в 100 раз. Допустимые значения свойства unit: ‘см’, ‘м’, ‘км’. Организуйте эту проверку. Продемонстрируйте работу с классом.

```python
# Решение 1
class Length():
    def __init__(self,invalue,inunit):
        self.__value = invalue
        self.__unit = inunit
    @property    
    def unit(self):
        return self.__unit
    @unit.setter
    def unit(self, new_unit=None):
        units = ["см","м","км"]
        if not new_unit:return self.__unit
        if units.index(new_unit) > units.index(self.__unit):
            self.__value /= 100 **  (units.index(new_unit) - units.index(self.__unit))
        elif  units.index(new_unit) < units.index(self.__unit):
            self.__value *= 100 **  (units.index(self.__unit) - units.index(new_unit))
        self.__unit = new_unit
    @property
    def value(self):
        return self.__value
    @value.setter
    def value(self, new_value=None):
        if not new_value:return self.__value
        self.__value = new_value
    
a = Length(100, "см")
print(a.value, a.unit)
a.unit = 'м'
print(a.value, a.unit)
a.unit = 'км'
print(a.value, a.unit)
a.unit = 'см'
print(a.value, a.unit)
a.value = 200
print(a.value, a.unit)
```

```python
# Решение 2.
class checklist():
    disciplines_list = ["Math", "IT", "Chemistry"]
    def __init__(self):
        self.__checklist = {}
    @property
    def discipline(self):
        return self.__discipline

    @discipline.setter
    def discipline(self, dcpl: str):
        if dcpl in self.disciplines_list:
            self.__discipline = dcpl

    @property
    def group(self):
        return self.__group

    @group.setter
    def group(self, grp: str):
        self.__group = grp

    def put(self,student: str, mark: str) -> None:
        self.__checklist[student] = mark
    def get(self, student: str) -> str:
        return self.__checklist[student]
    def change(self, student: str, mark: str) -> None:
        self.__checklist[student] = mark
    def delete(self, student) -> None:
        self.__checklist.pop(student)
    def result(self):
        res = [0,0,0,0,0]
        for key, value in self.__checklist.items():
            if value == "отлично": res[0] += 1
            elif value == "хорошо": res[1] += 1
            elif value == "удовл.":res[2] += 1
            elif value == "неудовл.":res[3] += 1
            elif value == "н/я": res[4] += 1
        return res
    def __str__(self):
        return str(self.__discipline +" " + self.__group + str([(k,v) for k,v in self.__checklist.items()]))

CList = checklist()
CList.discipline = "Math"
CList.group = "ID23-3"
CList.put("Ivan", "Отлично")
CList.put("Sasha", "Хорошо")
CList.put("Vasya", "удовл.")
print(CList.result())
print(CList)
CList.delete("Sasha")
print(CList)
```

4. Написать функцию my_func_1(lst), которая переворачивает список вложенных списков с произвольным уровнем вложенности.
Любой элемент, не являющийся списком, не должен быть изменен.

Примеры, демонстрирующие ожидаемое поведение функции:

assert my_func_1(42) == 42

assert my_func_1([]) == []

assert my_func_1([1, 2]) == [2, 1]

assert my_func_1((1, 2)) == (1, 2)

assert my_func_1([[1, 2], [3, [4, 5]], 6]) == [6, [[5, 4], 3], [2, 1]]

assert my_func_1(['123', '456', '789', {10: [11, 12], 13: 14}, (15, 16)]) == [(15, 16), {10: [11, 12], 13: 14}, '789', '456', '123']

assert my_func_1([[[[[1, 2]]]]]) == [[[[[2, 1]]]]]

```python
# Решение 4.
def my_func_1(lst: list) -> list:
    if type(lst)== list:
        l1 = lst.copy()
        for i in range(len(lst)):
            x = my_func_1(lst[i])
            l1[len(lst) - i - 1] = x
        return l1
    return lst


assert my_func_1(42) == 42

assert my_func_1([]) == []

assert my_func_1([1, 2]) == [2, 1]

assert my_func_1((1, 2)) == (1, 2)

assert my_func_1([[1, 2], [3, [4, 5]], 6]) == [6, [[5, 4], 3], [2, 1]]

assert my_func_1(['123', '456', '789', {10: [11, 12], 13: 14}, (15, 16)]) == [(15, 16), {10: [11, 12], 13: 14}, '789', '456', '123']

assert my_func_1([[[[[1, 2]]]]]) == [[[[[2, 1]]]]]
```

```python
# Решение 5.
print(sum(list(filter(lambda x: x % 9 != 0, map(lambda x: x**2, range(1,100))))))
```
