Класс – отрезок. Методы – расчет середины и длины. Поля класса – координаты начала, конца, середины и длина.

```python
class Segment():
    def __init__(self, start, end):
        self.start = start
        self.end = end
    def length(self):
        return self.end - self.start
    def middle(self):   
        return (self.end + self.start) / 2
a = Segment(2, 5)
print(f"Начало: {a.start}, конец {a.end}")
print(f"Длина: {a.length()}, середина {a.middle()}")
```

Опишите класс Angle, заданный величиной в градусах и минутах (двумя полями). Включите в описание класса методы, позволяющие реализовать перевод в радианы, привести величину угла к диапазону 0–360°, увеличить и уменьшить угол на заданную величину и свойство, позволяющее определить, является ли угол тупым.

```python
from math import radians


class Angle(object):
    def __init__(self, degrees: int, minutes: int) -> None:
        self.degrees = degrees
        self.minutes = minutes

    def radian(self) -> float:
        return radians(self.degrees + self.minutes / 60)

    def bring_to_0_360(self) -> float:
        degrees_plus_minutes = self.degrees + (self.minutes / 60)
        while degrees_plus_minutes > 360:
            degrees_plus_minutes -= 360
        while degrees_plus_minutes < 0:
            degrees_plus_minutes += 360
        return degrees_plus_minutes

    def cut(self, ndgrs: int, nmins: int = 0) -> None:
        self.degrees -= ndgrs
        if self.minutes - nmins < 0:
            self.degrees = (self.minutes - nmins) // 60
        self.minutes -= abs(nmins % 60)

    def enlarge(self, ndgrs: int, nmins: int = 0) -> None:
        self.degrees += ndgrs + (self.minutes + nmins) // 60
        self.minutes = (self.minutes + nmins) % 60
        
    @property
    def type(self) -> str:
        if 90 < self.bring_to_0_360() < 180:
            return "Угол тупой"
        else:
            return "Угол НЕ тупой"


if __name__ == "__main__":
    b = Angle(180, 50)
    print("Градусов", b.degrees, "минут", b.minutes)
    print("Угол в 0-360:", b.bring_to_0_360())
    print("Радиан", b.radian())
    print(b.type, '\n')
    b.enlarge(310, 40)
    print("Градусов", b.degrees, "минут", b.minutes)
    print("Угол в 0-360:", b.bring_to_0_360())
    print("Радиан", b.radian())
    print(b.type)
```
